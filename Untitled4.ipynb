```python
import pandas as pd
import matplotlib.pyplot as plt

# Example MD simulation data
time_ps = [0, 1, 2, 3, 4]
temperature_K = [295, 302, 298, 305, 300]
potential_energy = [-1200, -1205, -1202, -1207, -1204]

# Create a table
df = pd.DataFrame({
    "Time_ps": time_ps,
    "Temperature_K": temperature_K,
    "Potential_Energy": potential_energy
})

display(df)

# Plot temperature
plt.plot(time_ps, temperature_K, marker="o")
plt.xlabel("Time (ps)")
plt.ylabel("Temperature (K)")
plt.title("MD Simulation: Temperature vs Time")
plt.grid(True)
plt.show()
```
